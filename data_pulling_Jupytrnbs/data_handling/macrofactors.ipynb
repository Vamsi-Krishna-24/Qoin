```python
import pandas as pd

# ============================================
# QOIN — India CPI & WPI Monthly Data (From 2010)
# Source: FRED
# ============================================

START_DATE = "2010-01-01"

series = {
    # CPI Index (base 2015=100) — monthly, available up to recent
    "CPI_Index": "INDCPIALLMINMEI",
    # CPI YoY Inflation % — monthly
    "CPI_YoY_Inflation_%": "INDCPALTT01GYM",
    # WPI Index — monthly (FRED data ends Dec 2022, see note below)
    "WPI_Index": "WPIATT01INM661N",
}

df = pd.DataFrame()

for name, code in series.items():
    try:
        url = f"https://fred.stlouisfed.org/graph/fredgraph.csv?id={code}&cosd={START_DATE}"
        temp = pd.read_csv(url)
        temp.columns = ["Date", name]
        temp["Date"] = pd.to_datetime(temp["Date"])
        temp = temp.set_index("Date")
        temp[name] = pd.to_numeric(temp[name], errors="coerce")
        df = pd.concat([df, temp], axis=1)
        valid = temp[name].dropna()
        print(f"✅ {name}: {len(valid)} months ({valid.index[0].date()} to {valid.index[-1].date()})")
    except Exception as e:
        print(f"❌ {name}: Failed - {e}")

df = df.sort_index()
df.to_csv("india_cpi_wpi_monthly.csv")

print(f"\n💾 Saved: india_cpi_wpi_monthly.csv")
print(f"\nLast 12 months of CPI data:")
print(df[["CPI_Index", "CPI_YoY_Inflation_%"]].dropna().tail(12).round(2).to_string())

print(f"\nLast 12 months of WPI data:")
print(df[["WPI_Index"]].dropna().tail(12).round(2).to_string())

print("""
============================================================
NOTE ON WPI:
FRED's WPI series (WPIATT01INM661N) ends at Dec 2022.
For 2023-2026 WPI data, download from:
  → https://eaindustry.nic.in/download_data_1112.asp
  (Office of Economic Adviser, Govt of India)
  → Select 'All Commodities', download Excel, parse with pandas
============================================================
""")
```
