```python
import pandas as pd

# Load
s1 = pd.read_csv("/Users/surisettivamsikrishna/Downloads/Vamsi Pc/CODES/Mark4/Qoin/Qoin/Uncleaned_data/fii_dii.csv")
s2 = pd.read_csv("/Users/surisettivamsikrishna/Downloads/Vamsi Pc/CODES/Mark4/Qoin/Qoin/Uncleaned_data/fii_dii.csv", header=None, skiprows=2,
                 names=["Date","FII_Buy","FII_Sell","FII_Net",
                        "DII_Buy","DII_Sell","DII_Net","NIFTY_raw"])

# Standardize sheet 1
s1.columns = ["Date","FII_Buy","FII_Sell","FII_Net",
              "DII_Buy","DII_Sell","DII_Net"]
s1["Date"] = pd.to_datetime(s1["Date"])

# Standardize sheet 2
s2["Date"] = pd.to_datetime(s2["Date"], format="%d %b, %Y")

# Split NIFTY "23997.55-180.1 (-0.74%)"
nifty = s2["NIFTY_raw"].str.extract(
    r"([\d.]+)([-+]?[\d.]+)\s*\(([-+]?[\d.]+)%\)")
s2["Nifty_Close"]  = pd.to_numeric(nifty[0])
s2["Nifty_Change"] = pd.to_numeric(nifty[1])
s2["Nifty_%Chg"]   = pd.to_numeric(nifty[2])
s2 = s2.drop(columns=["NIFTY_raw"])

# Merge: prefer sheet 2 on overlaps
combined = (pd.concat([s1, s2])
              .drop_duplicates(subset="Date", keep="last")
              .sort_values("Date")
              .reset_index(drop=True))

combined.to_csv("fii_dii_2018_to_2026.csv", index=False)
```
